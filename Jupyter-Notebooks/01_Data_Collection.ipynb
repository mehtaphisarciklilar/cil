# Load and Inspect Data

## Load and inspect `house_prices_records` data.

Load data

```python
df = pd.read_csv(f"inputs/datasets/raw/house-price-20211124T154130Z-001/house-price/house_prices_records.csv")
df.head()
```

Inspect data

```python
df.info()
```

* There is information for the sale price of 1,460 properties in the data. 
* The data is not balanced - there are indicators with missing observations. Some of these may be at acceptable levels while some have very few number observations such as `EnclosedPorch` or `WoodDeckSF`.
* Missing observations for some features could be imputed by making some assumptions. For example, second floor squarefeet is possibly zero for the missing observations.  


Check if there are duplicate entries in data

```python
df[df.duplicated()]
```

There are no duplicate entries.
